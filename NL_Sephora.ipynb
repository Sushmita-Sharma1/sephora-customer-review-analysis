#**About Sephora:**
Sephora is one of the world's leading prestige beauty retailers, offering a wide range of products across skincare, makeup, haircare, fragrance, and wellness.
- Sephora operates thousands of physical stores and a global e-commerce platform, serving millions of customers worldwide.
- The company partners with both luxury and emerging beauty brands to provide customers with a diverse selection of high-quality products.

#**Objective**

- The analysis focuses on understanding how factors such as brand, product category, price, customer reviews, product attributes, and customer demographics affect customer satisfaction. By uncovering these insights, the project aims to support data-driven decisions that improve customer experience, strengthen brand loyalty, and increase repeat purchases.

#**Business Problem Overview**

- Identify the brands and products with the highest and lowest customer satisfaction.
- Analyze the relationship between product price and customer ratings.
Evaluate how customer engagement metrics (reviews, recommendations, helpful votes) influence satisfaction.
- Understand the impact of product attributes such as limited edition, new launches, Sephora exclusive, online-only, and product variants on customer ratings.
- Provide actionable recommendations to help Sephora optimize product assortment, pricing strategy, and customer experience.

Premium products receive the highest average rating (4.34), but the difference compared with Budget and Mid-priced products (4.31) is minimal. #Which products get poor ratings and negative reviews?

- Lip Sleeping Mask Intense Hydration with Vitamin C is 1.47 poor rating and high negative review(1859)
- Protini Polypeptide Firming Refillable Moisturizer	1.44 rating and negative review is	1203	etc.

### Key Finding

Analysis of low-rated products reveals that customer dissatisfaction is primarily associated with product performance rather than price or brand. The most common complaints involve skin irritation, breakouts, dryness, heavy or sticky texture, fragrance sensitivity, and products not meeting customer expectations.

### Business Implication

Mining customer review text alongside ratings enables Sephora to identify recurring product issues, prioritize quality improvements, and make data-driven decisions that enhance customer satisfaction and reduce product returns.

````python
negative_reviews = df[df['rating_x'] <= 2]
negative_product = (
    negative_reviews
    .groupby('product_name_x')
    .agg(
        avg_rating=('rating_x','mean'),
        negative_review_count=('total_neg_feedback_count','count'),
        example_review=('review_text','first')
    )
    .sort_values(
        'negative_review_count',
        ascending=False
    )
)

negative_product.head(20)
````

The analysis identified several products with both exceptionally low average ratings (1.35–1.52 stars) and a high volume of negative reviews. The Lip Sleeping Mask Intense Hydration with Vitamin C (1,859 negative reviews) and Protini Polypeptide Firming Refillable Moisturizer (1,203 negative reviews) emerged as the most critical products requiring immediate attention. Additionally, identical customer sentiment across mini and full-size variants suggests that dissatisfaction is linked to product performance rather than packaging, highlighting opportunities for product reformulation and quality improvements.

````python
plt.figure(figsize=(10, 6))
sns.barplot(data=negative_product.head(10), x='avg_rating', y='product_name_x')
plt.title('Top 10 Products with Lowest Average Rating')
plt.xlabel('Average Rating')
plt.ylabel('Product Name')
plt.show()
````

The scatter plot reveals that most low-rated products have average ratings between 1.4 and 1.6 stars. However, a small number of products generate a disproportionately high volume of negative reviews, making them critical candidates for quality improvement. The analysis also shows that average rating alone does not fully explain customer dissatisfaction, highlighting the importance of considering both review volume and rating when prioritizing product improvements.

````python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=negative_product, x='avg_rating', y='negative_review_count')
plt.title('Average Rating vs. Negative Review Count for Low-Rated Products')
plt.xlabel('Average Rating')
plt.ylabel('Negative Review Count')
plt.show()
````
